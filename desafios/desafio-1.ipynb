Primeiro desafio:

Você é gerente do banco BANCOESTADO e precisa realizar a gestão e carga de ATM's. Um caixa eletrônico precisa ser abastecido semanalmente com cassetes de cédulas de 10,00, 20,00, 50,00 e 100,00. Cada cassete é preenchido com 2500 cédulas, e você terá um parque de 7 ATM's para gerenciar. Todos os ATM's devem possuir o mesmo saldo, você só pode preencher os cassetes com as cédulas disponíveis ( 10, 20, 50 100 )

1 - Você recebeu um malote de R$ 3.298.240,00 e precisa abastecer todos os caixas eletrônicos com o mesmo saldo. Lembre-se das premissas citadas no enunciado acima.  Retorne o valor de abastecimento máximo possível de cada ATM. Os nomes dos ATM's devem ser nomeados com a sigla ATM mais o número do ATM.
- Retorne o valor que retornará para o cofre
- Retorne o saldo de cada caixa eletrônico
- Retorne o quantitativo de notas por cassete

PS: O arquivo de saída esperado deverá ser nesse formato:

# Possível solução

````python
# Para cada ATM eu preciso armazenar um saldo de notas, a estrutura de dados utilizadas deve ser flexível ao ponto de pode gerenciar várias ATM's independente do número.
# Existem várias maneiras de armazenar esse tipo de dados, eu vou optar por guardar em dicionários. # https://docs.python.org/pt-br/3/tutorial/datastructures.html
ATMS = {
  "ATM1":{
      "cassete_10": 0,
      "cassete_20": 0,
      "cassete_50": 0,
      "cassete_100": 0,
      "saldo_atm": 0,
      },
  "ATM2":{
      "cassete_10": 0,
      "cassete_20": 0,
      "cassete_50": 0,
      "cassete_100": 0,
      "saldo_atm": 0,
      },
  "ATM3":{
      "cassete_10": 0,
      "cassete_20": 0,
      "cassete_50": 0,
      "cassete_100": 0,
      "saldo_atm": 0,
      },
  "ATM4":{
      "cassete_10": 0,
      "cassete_20": 0,
      "cassete_50": 0,
      "cassete_100": 0,
      "saldo_atm": 0,
      },
  "ATM5":{
      "cassete_10": 0,
      "cassete_20": 0,
      "cassete_50": 0,
      "cassete_100": 0,
      "saldo_atm": 0,
      },
  "ATM6":{
      "cassete_10": 0,
      "cassete_20": 0,
      "cassete_50": 0,
      "cassete_100": 0,
      "saldo_atm": 0,
      },
  "ATM7":{
      "cassete_10": 0,
      "cassete_20": 0,
      "cassete_50": 0,
      "cassete_100": 0,
      "saldo_atm": 0,
      },
}
````

````python
# Agora vamos verificar como ficou o nosso dicionário.
print(ATMS)
````

````python
# Agora vamos trabalhar com o malote que chegou: R$ 3.298.240,00
# Para operações financeiras, você deve evitar ao máximo utilizar pontos flutuantes, como float. Pois ele não é muito confiável para divisões e arredondamentos.
# Para esse desafio, vamos optar por utilizar apenas 2 casas decimais e preenchermos com zero, logo o malote de 3.298.240,00 seria 329824000
FATOR_DE_CONVERSAO = 100                                                        # Esse fator fará a conversão de reais para centavos de reais.
MALOTE = int(3298240.00 * FATOR_DE_CONVERSAO)


# estaremos operando nos centavos
print(MALOTE)
````

````python
# Agora eu vou calcular o saldo máximo dos ATMS, para isso eu preciso calcular o total de valores em cada cassete, sendo que cada cassete possui 2500 notas.
# 2500 notas * 10 reais =
# 2500 notas * 20 reais =
# 2500 notas * 50 reais =
# 2500 notas * 100 reais =

NUMERO_MAXIMO_NOTAS = 2500

# Lembre-se que estamos operando na casa dos centavos, no final da expressão eu estou convertendo em centavos
SALDO_MAXIMO_POR_ATM = ((NUMERO_MAXIMO_NOTAS * 10) + (NUMERO_MAXIMO_NOTAS * 20) + (NUMERO_MAXIMO_NOTAS * 50) + (NUMERO_MAXIMO_NOTAS * 100)) * FATOR_DE_CONVERSAO

# Saldo máximo por ATM é de 45000000
print("Saldo máximo por ATM é de", SALDO_MAXIMO_POR_ATM, "centavos")

# Saldo total dos ATM's será de
QUANTIDADEDE_ATMS = 7
SALDO_TOTAL_ATMS = SALDO_MAXIMO_POR_ATM * QUANTIDADEDE_ATMS
print("Saldo total dos ATM's será de", SALDO_TOTAL_ATMS, "centavos")
````

````python
# Primeiro passo é verificar se o meu malote tem dinheiro suficiente para abastecer todos os atms.
if MALOTE >= SALDO_TOTAL_ATMS:
  print("O malote tem dinheiro suficiente para abastecer todos os atms")
else:
  print("O malote não tem dinheiro suficiente para abastecer todos os atms")
````

````python
# Retorne o valor que retornará para o cofre:

SALDO_DO_COFRE = MALOTE - SALDO_TOTAL_ATMS
print("Retorne o valor que retornará para o cofre:", SALDO_DO_COFRE, "centavos")
````

````python
def adicionar_saldo_em_atm(ATM_ID:str, CASSETE:str, SALDO:int ):

  ATMS[ATM_ID][CASSETE] += SALDO
  TOTAL = (ATMS[ATM_ID]["cassete_10"] * 10 + ATMS[ATM_ID]["cassete_20"] * 20 + ATMS[ATM_ID]["cassete_50"] * 50 + ATMS[ATM_ID]["cassete_100"] * 100) * FATOR_DE_CONVERSAO
  ATMS[ATM_ID]["saldo_atm"] = TOTAL
  return ATMS

#adicionar_saldo_em_atm("ATM1", "cassete_10", 250000)
#adicionar_saldo_em_atm("ATM1", "cassete_20", 250000)
#adicionar_saldo_em_atm("ATM1", "cassete_50", 250000)
#adicionar_saldo_em_atm("ATM1", "cassete_100", 250000)
````

````python
# Retorne o saldo de cada caixa eletrônico:
for ID, i in ATMS.items():
  #CONT = 0
  for j in i:
    adicionar_saldo_em_atm(ID, j, 250000)
  print(f'{ID}: {ATMS[ID].get("saldo_atm")}')
  ATMS[ID].popitem()
# Retorne o quantitativo de notas por cassete:
  print(f'{ATMS[ID]}\n')
````

# Solução

````python
malote = 3298240
notas = [100,50,20,10]
cedulas_por_atm = [0,0,0,0]
total_dinheiro = 0
valor_individual = 0
p = malote // 7

for i in range(len(notas)):
  if (p // notas[i] > 0):
    cedulas_por_atm[i] = p//notas[i]
    p = p % notas[i]

if (sum(cedulas_por_atm) > 2500):
  valor_individual = 2500
  cedulas_por_atm[0] = valor_individual
  cedulas_por_atm[1] = cedulas_por_atm[2] = cedulas_por_atm[3] = 0
else:
  valor_individual = sum(cedulas_por_atm)

for i in range(len(notas)):
  total_dinheiro += cedulas_por_atm[i]*notas[i]
total_dinheiro = (total_dinheiro*7)
print("{\n"
      "  'sucesso': true,\n"
      "  'distribuicao_por_atm': {\n"
     f"    'valor_individual': {valor_individual}\n"
      "    'cedulas_por_atm:' {"
)

for i in range(len(cedulas_por_atm)):
  print(f"      R$'{notas[i]}': {cedulas_por_atm[i]}")

print("    } \n  },\n"
      "  'atms_afetados': ['ATM1', 'ATM2', 'ATM3', 'ATM4', 'ATM5', 'ATM6', 'ATM7'],\n"
    f"  'sobra_cofre_central': {malote - (total_dinheiro)}\n"
    "}"
    )
````
