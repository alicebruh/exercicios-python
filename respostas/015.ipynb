```python
g = float(input("Quantidade ganha por hora: "))
h = float(input("Quantidade de horas trabalhadas no mês: "))
sb = g*h
ir = 0.11*sb
inss = 0.08*sb
s = 0.05*sb
sl = sb - (ir+inss+s)
print(f"+ Salário Bruto : R${sb:.2f}\n"
      f"- IR (11%) : R${ir:.2f}\n"
      f"- INSS (8%) : R${inss:.2f}\n"
      f"- Sindicato (5%) : R${s:.2f}\n"
      f"= Salário Liquido : R${sl:.2f}\n"
      )
```
