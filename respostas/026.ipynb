```python
p1 = float(input("Digite o preço do primeiro produto: "))
p2 = float(input("Digite o preço do segundoo produto: "))
p3 = float(input("Digite o preço do terceiro produto: "))

if (p1 < p2 and p1 < p3):
  mn = p1
elif(p2 < p1 and p2 < p3):
  mn = p2
else:
  mn = p3
print(f"Você deve comprar o produto de preço R${mn:.2f}")
```
