```python
n = int(input("Digite a quantidade de pessoas: "))
s = 0
for i in range(n):
  x = int(input("Digite as idades das pessoas: "))
  s += x
m = s/n
if (m <= 25):
  print(f"Em média, esse grupo de pessoas é jovem")
elif (m > 60):
  print(f"Em média, esse grupo de pessoas é idoso")
else:
  print(f"Em média, esse grupo de pessoas é adulto")
```
