```python
n = int(input("Digite a quantidade de números no conjunto: "))
c = []
s = 0
mn = 1000
mx = 0

for i in range(n):
  x = int(input("Digite os números presentes no conjunto: "))
  while x < 0 or x > 1000:
    x = int(input("Número inserido é maior ou menor que o limite do programa, por favor digite um número válido: "))
  c.append(x)

for i in c:
  if (i > mx):
    mx = i
  if (i < mn):
    mn = i
  s += i
print(f"O menor valor do conjunto é {mn}, o maior valor do conjunto é {mx}, e a soma dos valores dos conjuntos é {s}.")
```
