```python
n = int(input("Digite um número inteiro: "))
p = True
for i in range(n):
  if ((i+1 != 1 and i+1 != n) and n % (i+1) == 0):
    p = False
    print(f"{n} não é primo e é divisível por {i+1}")
if p:
  print(f"{n} é um número primo")
```
