```python
v = []
c = 0
for i in range(10):
  v.append(input(f'Digite o caractere número {i+1}: ').upper())
  if (v[i] != 'A' and v[i] != 'E' and v[i] != 'I' and v[i] != 'O' and v[i] != 'U'):
    c += 1
print(f"Esse vetor tem {c} consoantes.\n"
       "As consoantes nesse vetor são:", end="")
for i in v:
  if (i != 'A' and i != 'E' and i != 'I' and i != 'O' and i != 'U'):
    print(f" {i}", end="")
```
