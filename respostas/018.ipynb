```python
a = float(input("Tamanho do arquivo, em MB: "))
v = float(input("Velocidade da internet, em Mbps: "))
m = (a/v)//60
s = (a/v)%60
print(f"O tempo aproximado de download do arquivo é {m:.0f} minutos e {s:.0f} segundos")
```
