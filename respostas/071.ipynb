```python
t = int(input("Quantidade de turmas: "))
s = 0
for i in range(t):
  a = int(input("Quantidade de alunos em cada turma: "))
  while a > 40:
    a = int(input("Quantidade inválida inserida! Por favor insera o número correto de alunos: "))
  s += a
print(f"A média de alunos em cada turma é {(s/t):.2f}")
```
