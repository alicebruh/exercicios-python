```python
x = 0
a = float(input("Digite a quantidade de habitantes do país A: "))
b = float(input("Digite a quantidade de habitantes do país B: "))
t1 = float(input("Digite a taxa de crecimento do país A: "))
t2 = float(input("Digite a taxa de crecimento do país B: "))

while (a < b):
  x += 1
  a = a * ((1 + (t1/100))**x)
  b = b * ((1 + (t2/100))**x)

if (a == b):
  print(f"No ano {x}, as populações dos países A e B se igualaram.")
else:
  print(f"No ano {x}, a população do países A superou a população do país B.")
```
