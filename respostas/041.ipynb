```python
n1 = float(input("Primeiro número: "))
n2 = float(input("Segundo número: "))
o = input("Deseja fazer que tipo de operação? ")

if (o == '+'):
  r = n1+ n2
elif (o == '-'):
  r = n1 - n2
elif (o == '*'):
  r = n1*n2
elif (o == '/'):
  r = n1/n2
else:
  r = 0

if (r >= 0):
  t1 = "positivo"
else:
  t1 = "negativo"

if (r % 2 == 0):
  t2 = "par"
else:
  t2 = "ímpar"

if (r % 1 == 0):
  t3 = "inteiro"
else:
  t3 = "decimal"

print(f"O resultado da operação é {r}, e ele é: {t1}, {t2} e {t3}.")
```
