```python
v1 = []
v2 = []
v3 = []

for i in range(10):
  v1.append(input(f"Digite o elementos {i+1} do primeiro vetor: "))

for i in range(10):
  v2.append(input(f"Digite os elementos {i+1} do segundo vetor: "))

for i in range(10):
  v3.append(v1[i])
  v3.append(v2[i])

print(f"Terceiro vetor:", end="")
for i in v3:
  print(f" {i}", end="")
```
