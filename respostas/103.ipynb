```python
idades = []
alturas = []
x = 0

for i in range(30):
  idades.append(int(input(f"Idade do aluno {i+1}: ")))
  alturas.append(int(input(f"Altura do aluno {i+1}, em cm: ")))

for i in idades:
  if (i > 13 and alturas[i] < (sum(alturas))/30):
    x += 1
print(f"A quantidade de alunos com idade maior que 13 e altura menor que a média {x}")
```
