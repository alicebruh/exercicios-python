```python
x = float(input("Primeiro número: "))
y = float(input("Segundo número: "))
z = float(input("Terceiro número: "))

if (x > y and x > z):
  mx = x
  if (y > z):
    mn = z
    m = y
  else:
    mn = y
    m = z
elif(y > x and y > z):
  mx = y
  if (x > z):
    mn = z
    m = x
  else:
    mn = x
    m = z
else:
  mx = z
  if (x > y):
    mn = y
    m = x
  else:
    mn = x
    m = y
print(f"A ordem dos números é: {mx:.2f}, {m:.2f}, e {mn:.2f}")
```
