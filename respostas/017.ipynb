```python
import math
a = float(input("Tamanho da área a ser pintada: "))

litro = a/6
l1 = math.ceil(litro/18)
g1 = math.ceil(litro/3.6)

l2 = math.floor(litro/18)
g2 = math.ceil((litro%18)/3.6)

print(f"Quantidade de latas para pintar a área: {int(l1)}\n"
      f"Valor total: R${(l1*80):.2f}\n"
      f"Quantidade de galões para pintar a área: {int(g1)}\n"
      f"Valor total: R${(g1*25):.2f}\n"
      f"Quantidade de latas e galões para pintar a área: {int(l2)} latas e {int(g2)} galões\n"
      f"Valor total: R${((l2*80)+(g2*25)):.2f}")
```
