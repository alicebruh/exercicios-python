```python
v = []
soma_quadrados = 0

for i in range(10):
  v.append(int(input("Digite um número inteiro: ")))
  soma_quadrados += v[i]**2
print(f"A soma dos quadrados dos números dentro do vetor é igual a {soma_quadrados}")
```
