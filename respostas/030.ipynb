```python
v = float(input("Valor da hora: "))
h = float(input("Número de horas: "))
sb = v * h
if (sb <= 900):
  ir = 0
elif(sb <= 1500):
  ir = 5
elif(sb <= 2500):
  ir = 10
else:
  ir = 20
print(f" Salário Bruto: ({v:.2f} * {h:.2f})        : R$ {sb:.2f}\n"
      f" (-) IR ({ir}%)                           : R$ {(ir/100)*sb:.2f}\n"
      f" (-) INSS (10%)                        : R$ {0.10*sb:.2f}\n"
      f" FGTS (11%)                            : R$ {0.11*sb:.2f}\n"
      f" Total de descontos                    : R$ {(0.10*sb)+((ir/100)*sb):.2f}\n"
      f" Salário Liquido                       : R$ {(sb - ((0.10*sb)+((ir/100)*sb))):.2f}")
```
