```python
a = int(input("Digite um ano: "))
if (a % 4 == 0):
  if (a % 100 == 0):
    if (a % 400 == 0):
      print("O ano é bissexto")
    else:
      print("O ano não é bissexto")
  else:
    print("O ano é bissexto")
else:
  print("O ano não é bissexto")
```
