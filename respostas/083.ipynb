```python
i1= i2 = i3 = i4 = 0

while True:
  n = float(input("Digite um número: "))
  if (n < 0):
    break
  if(n <= 100 and n >= 76):
    i1 += 1
  elif(n >= 51):
    i2 += 1
  elif(n >= 26):
    i3 += 1
  else:
    i4 += 1
print(f"A quantidade de números no intervalo [0-25] é {i4};\n"
      f"A quantidade de números no intervalo [26-50] é {i3};\n"
      f"A quantidade de números no intervalo [51-75] é {i2};\n"
      f"A quantidade de números no intervalo [76-100] é {i1}.")
```
