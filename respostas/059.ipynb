```python
l = []
p = 0
im = 0

for i in range(10):
  n = int(input("Digite 10 números inteiros: "))
  l.append(n)

for i in l:
  if (i % 2 == 0):
    p += 1
  else:
    im += 1
print(f"A quantidade de números ímpares é {im} e a de números pares é {p}")
```
