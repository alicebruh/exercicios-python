```python
n = int(input("Montar a tabuada de: "))
x = int(input("Começar por: "))
y = int(input("Terminar em: "))

while y < x:
  y = int(input("Número inválido! Digite outro: "))

for i in range((y-x)+1):
  print(f"{n} X {x+i} = {n*(x+i)}")
```
