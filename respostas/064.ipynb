```python
while True:
  n = int(input("Digite o número o qual você quer saber o fatorial: "))
  while n > 16 or n < 0:
    n = int(input("O número digitado está fora dos parâmetros do programa, por favor digite outro número: "))
  x = 1
  for i in range(n):
    x *= n-i
  print(f"O fatorial de {n} é {x}")
  c = input("Deseja continuar a usar o programa (S/N)? ")
  if (c != 'S'):
    break
```
