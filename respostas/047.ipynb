```python
n = input("Nome de usuário: ")
s = input("Senha do usuário: ")

while (n == s):
  n = input("Erro! Insira o nome novamente: ")
  s = input("Insira a senha novamente: ")
print("Nome e senha de usuário válidos!")
```
