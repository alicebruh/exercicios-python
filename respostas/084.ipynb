```python
p1 = p2 = p3 = p4 = p5 = p6 = 0

while True:
  pedido = int(input("Especificação do pedido: "))
  if (pedido < 100 or pedido > 105):
    break
  if (pedido == 100):
    p1 += 1
  elif (pedido == 101):
    p2 += 1
  elif (pedido == 102):
    p3 += 1
  elif (pedido == 103):
    p4 += 1
  elif (pedido == 104):
    p5 += 1
  else:
    p6 += 1
print("Codigo  |  Quantidade  |  Preço unidade  |  Preço total\n"
      f"100       {p1}               R$ 1.20           R${(p1*1.2):.2f}\n"
      f"101       {p2}               R$ 1.30           R${(p2*1.3):.2f}\n"
      f"102       {p3}               R$ 1.50           R${(p3*1.5):.2f}\n"
      f"103       {p4}               R$ 1.20           R${(p4*1.2):.2f}\n"
      f"104       {p5}               R$ 1.30           R${(p5*1.3):.2f}\n"
      f"106       {p6}               R$ 1.00           R${(p6*1.0):.2f}\n"
      f"Total do Pedido: R${((p1*1.2)+(p2*1.3)+(p3*1.5)+(p4*1.2)+(p5*1.3)+(p6*1.0)):.2f}")
```
