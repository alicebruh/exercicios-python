```python
from math import inf
mn = inf
mx = -inf
s = 0
x = 0
while True:
  t = float(input("Insira uma temperatura, em Kelvin, e insira ela negativa para parar o programa: "))
  if (t >= 0):
    s += t
    x += 1
    if (mn > t):
      mn = t
    if (mx < t):
      mx = t
  else:
    break

print(f"A maior temperatura do conjunto é {mx}, a menor é {mn} e a média das temperaturas é {(s/x):.2f}")
```
