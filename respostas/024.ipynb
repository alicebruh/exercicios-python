```python
x = float(input("Digite um número: "))
y = float(input("Digite outro número: "))
z = float(input("Digite mais outro número: "))

if (x > y and x > z):
  m = x
elif (y > x and y > z):
  m = y
else:
  m = z
print(f"O maior número, entre os três digitados, é {m:.2f}")
```
