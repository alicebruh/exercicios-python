```python
d = int(input("Digite um número: "))
if (d <= 7 and d > 0):
  if (d == 1):
    print("Domingo")
  elif (d == 2):
    print("Segunda")
  elif (d == 3):
    print("Terça")
  elif (d == 4):
    print("Quarta")
  elif (d == 5):
    print("Quinta")
  elif (d == 6):
    print("Sexta")
  else:
    print("Sábado")
else:
  print("Valor inválido")
```
