```python

while True:
  n = input("Digite o nome do atleta ou 0 encerrar programa: ")
  if (n == '0'):
    break
  atletas = {
    "nome": n,
    "notas": [],
    "melhor_nota": 0,
    "pior_nota": 0,
    "media": 0
  }

  for i in range(7):
    atletas["notas"].append(float(input(f"Nota número {i+1} do atleta: ")))
  atletas["notas"].sort()
  atletas["media"] = (atletas["notas"][1]+atletas["notas"][2]+atletas["notas"][3]+atletas["notas"][4]+atletas["notas"][5])/5
  print("Resultado Final:\n"
        f"Atleta: {atletas["nome"]}\n"
        f"Melhor nota: {atletas["notas"][6]}\n"
        f"Pior nota: {atletas["notas"][0]}\n"
        f"Média: {atletas["media"]}")
```
