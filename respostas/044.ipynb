```python
x = float(input("Quantidade de morango, em Kg: "))
y = float(input("Quantidade de maçãs em Kg: "))
#preço: morango
if (x <= 5):
  px = 2.50*x
else:
  px = 2.20*x
#preço: maçã
if (y <= 5):
  py = 1.80*x
else:
  py = 1.5*x

p = py+px
if (x+y > 8 or p > 25):
  p = p *0.9

print(f"O valor a ser pago pelo cliente é R${p:.2f}")
```
