```python
import math

a = float(input("Valor do a da equação: "))
if (a != 0):
  b = float(input("Valor do b da equação: "))
  c = float(input("Valor do c da equação: "))
  d = (b**2) - 4*a*c
  if (d > 0):
      x1 = (-b + math.sqrt(d))/(2*a)
      x2 = (-b - math.sqrt(d))/(2*a)
      print(f"As raízes da equação são {x1:.2f} e {x2:.2f}")
  elif(d == 0):
      x = -b/(2*a)
      print(f"A raiz quadrada dessa equação é {x:.2f}")

  else:
    print("A equação não tem uma raiz real")

else:
  print("A equação não é de segundo grau")
```
