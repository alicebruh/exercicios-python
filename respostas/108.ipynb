```python
votos = []
num = [0,0,0,0,0,0]
sistemas = ['Windows Server','Unix','Linux','Netware','Mac OS','Outro']
mais_votos = 0
maior_porc = 0
mais_votado = ""

def calc_porc(y):
  return(y*100)/len(votos)

print("Qual o melhor Sistema Operacional para uso em servidores?\n"
      "(1- Windows Server 2- Unix 3- Linux 4- Netware 5- Mac OS 6- Outro)")
while True:
  x = int(input("Selecione um sistema operacional, ou digite 0 para parar o programa: "))
  if (x == 0):
    break
  elif(x > 6 or x < 0):
    print("Selecione um valor válido!")
  else:
    votos.append(x)

for i in range(6):
  num[i] = votos.count(i+1)
  if (num[i] > mais_votos):
    mais_votos = num[i]
    mais_votado = sistemas[i]
    maior_porc = calc_porc(num[i])

print(f"Sistema Operacional - Votos - %\n")
for i in range(len(num)):
  if(num[i] != 0):
    print(f"{sistemas[i]} - {num[i]} - {calc_porc(num[i]):.1f}%")

print(f"\nTotal: {len(votos)}\n")
print(f"O Sistema Operacional mais votado foi o {mais_votado}, com {mais_votos} votos, correspondendo a {maior_porc:.1f}% dos votos.")
```
