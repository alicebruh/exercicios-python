```python
n = int(input("Digite um número inteiro maior que zero e menor que 1000: "))
c = n // 100
d = (n % 100)//10
u = n % 10

x1 = ""
x2 = ""

if (c > 0):
  if (c > 1):
    t1 = f"{c} centenas"
  else:
    t1 = f"{c} centena"
else:
  t1 = ""

if (d > 0):
  if (d > 1):
    t2 = f"{d} dezenas"
  else:
    t2 = f"{d} dezena"
else:
  t2 = ""

if (u > 0):
  if (u > 1):
    t3 = f"{u} unidades"
  else:
    t3 = f"{u} unidade"
else:
  t3 = ""

if (c > 0 and d > 0 and u > 0):
  x1 = ", "
  x2 = " e "
elif((c > 0 and d > 0) or (c > 0 and u > 0)):
  x1 = " e "
elif (d > 0 and u > 0):
  x2 = " e "

print(f"{t1}{x1}{t2}{x2}{t3}")
```
