```python
s = float(input("Informe seu salário: R$"))
if (s <= 280):
  p = 20
elif (s <= 700):
  p = 15
elif (s <= 1500):
  p = 10
else:
  p = 5

v = (p/100)*s
n = v + s
print(f"Salário antes do reajuste: R${s:.2f}\n"
      f"Percentual de aumento aplicado: {p}%\n"
      f"Valor do aumento: R${v:.2f}\n"
      f"Novo salário, após o aumento: R${n:.2f}")
```
