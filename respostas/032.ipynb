```python
n1 = float(input("Primeira nota: "))
n2 = float(input("Segunda nota: "))
m = (n1+n2)/2
if (m > 9):
  c = 'A'
elif (m > 7.5):
  c = 'B'
elif (m > 6):
  c = 'C'
elif (m > 4):
  c = 'D'
else:
  c = 'E'

if (c == 'D' or c == 'E'):
  r = 'REPROVADO'
else:
  r = 'APROVADO'
print(f'Com as notas {n1:.2f} e {n2:.2f}, a média é {m:.2f} e o conceito é {c}\n'
      f'{r}')
```
