```python
from math import inf

maior_indice = -inf
menor_indice = inf
media_v = media_a = x = 0
c_maior = c_menor = ""

for i in range(5):
  c = input("Código da cidade: ")
  nv = int(input("Número de veículos de passeio (em 1999): "))
  na = int(input("Número de acidentes de trânsito com vítimas (em 1999): "))

  media_v += nv
  if (nv < 2000):
    media_a += na
    x = i+1

  if (na > maior_indice):
    maior_indice = na
    c_maior = c
  if (na < menor_indice):
    menor_indice = na
    c_menor = c

print(f"O maior índice de acidentes de transito é de {maior_indice} e pertence a cidade de código {c_maior};\n"
      f"O menor índice de acidentes de transito é de {menor_indice} e pertence a cidade de código {c_menor};\n"
      f"A média de veículos das 5 cidades é {(media_v/5):.2f}, e a média de de acidentes de trânsito nas cidades com menos de 2.000 veículos de passeio é {(media_a/x):.2f}.")
```
