```python
x = float(input("Digite um número: "))
y = float(input("Digite outro número: "))
z = float(input("Digite mais outro número: "))

if (x > y and x > z):
  mx = x
  if (y > z):
    mn = z
  else:
    mn = y
elif (y > x and y > z):
  mx = y
  if (x > z):
    mn = z
  else:
    mn = x
else:
  mx = z
  if (x > y):
    mn = y
  else:
    mx = x
print(f"Dos três números, o maior é {mx:.2f} e o menor é {mn:.2f}")
```
