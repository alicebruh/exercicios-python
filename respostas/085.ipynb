```python
c1 = c2 = c3 = c4 = b = n = 0
while True:
  v = int(input("Digite o código dos candidatos: "))
  if (v < 1):
    break
  if (v == 1):
    c1 += 1
  elif (v == 2):
    c2 += 1
  elif (v == 3):
    c3 += 1
  elif (v == 4):
    c4 += 1
  elif (v == 5):
    n += 1
  else:
    b += 1
print(f"Total de Votos para o Candidato 1: {c1}\n"
      f"Total de Votos para o Candidato 2: {c2}\n"
      f"Total de Votos para o Candidato 3: {c3}\n"
      f"Total de Votos Nulos: {n}\n"
      f"Total de Votos Brancos: {b}\n"
      f"Percentagem de votos nulos sobre o total de votos: {(100*n)/(c1+c2+c3+c4+b+n):.2f}%\n"
      f"Percentagem de votos brancos sobre o total de votos: {(100*b)/(c1+c2+c3+c4+b+n):.2f}%")
```
