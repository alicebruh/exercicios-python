```python
d = input("Digite uma data: ").split("/")

dia = int(d[0])
m = int(d[1])
a = int(d[2])

if (a > 0):
  if (m <= 12 and m > 0):
    if(m in [1, 3, 5, 7, 8, 10, 12]):
      if(dia > 0 and dia <= 31):
        print("Data Válida!")
      else:
        print("Dia Inválido!")
    elif (m in [4, 6, 9, 11]):
      if (dia > 0 and dia <= 30):
        print("Data Válida!")
      else:
        print("Dia Válida!")
    else:
      if (a % 4 == 0):
        if (dia > 0 and dia <= 30):
          print("Data Válida!")
        else:
          print("Dia Inválido")
      else:
        if (dia > 0 and dia <= 29):
          print("Data Válida!")
        else:
          print("Dia Inválido")
  else:
    print("Mês Inválido")
else:
  print("Ano Inválido")
```
