```python
notas = []
x = 0
y = 0
b = 0
while True:
  b = float(input("Digite um valor, ou -1 para encerrar o programa: "))
  if (b < 0):
    break
  notas.append(b)

print(f"Quantidade de notas lidas: {len(notas)}")
print(f"Notas na ordem que chegaram:", end="")
for i in notas:
  print(f" {i}", end="")
  if (i > sum(notas)/len(notas)):
    x += 1
  if (i < 7):
    y += 1

print(f"\nNotas na ordem inversa a que chegaram:", end="")

for i in range(len(notas)):
  print(f" {notas[len(notas)-(i+1)]}", end="")
print(f"\nSoma dos valores lidos: {sum(notas)}")
print(f"Média dos valores lidos: {(sum(notas)/len(notas)):.2f}")
print(f"Quantidade de valores acima da média: {x}")
print(f"Quantidade de valores abaixo de 7: {y}")
```
