```python
n = int(input("Digite o valor de n: "))
m = 0
s = 0
print(f"S = ", end="")
for i in range(n):
  print(f"{i+1}/{(i*2)+1} ",end="")
  if i < n-1:
    print("+ ",end="")

  s += (i+1)/((i*2)+1)
print(f"\nS = {s:.2f}")
```
