```python
medias = []
f = 0
for i in range(10):
  y = 0
  print(f"Aluno {i+1}:")
  for j in range(4):
    x = float(input(f"Nota número {j+1}: "))
    y += x
  medias.append((y/4))
  if medias[i] >= 7:
    f += 1
print(f"A quantidade de alunos com média igual ou maior que 7 é {f}")
```
