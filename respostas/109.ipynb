```python
salarios = []
abonos = []
valor_minimo = 0
valor_maior = 0

while True:
  x = int(input("Salário: "))
  if (x == 0):
    break
  salarios.append(x)
print("\nSalário - Abono\n")
for i in salarios:
  if (0.2*i > 100):
    abonos.append(0.2*i)
    print(f"R${i:.2f} - R${(0.2*i):.2f}")
    if (0.2*i > valor_maior):
      valor_maior = 0.2*i
  else:
    abonos.append(100)
    print(f"R${i:.2f} - R$100.00")
    valor_minimo += 1

print(f"\nForam processados {len(salarios)} colaboradores\n"
      f"Total gasto com abonos: R$ {sum(abonos):.2f}\n"
      f"Valor mínimo foi pago a {valor_minimo} colaboradores\n"
      f"Maior valor de abono pago: R${valor_maior:.2f}")
```
