```python
n1 = int(input("Digite um número inteiro: "))
n2 = int(input("Digite outro número inteiro: "))
if (n1 > n2):
  x = n1
  y = n2
else:
  x = n2
  y = n1
z = x - y
s = 0
for i in range(z):
  if(y < x-1 and y != x-1):
    y += 1
    s += y

print(s)
```
