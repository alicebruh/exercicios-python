```python
from math import inf
mn = inf
mx = -inf
nmn = nmx =""

for i in range(10):
  n = input("O número do aluno é: ")
  a = int(input("A altura do aluno, em centímetros, é: "))
  if (a > mx):
    mx = a
    nmx = n
  if (a < mn):
    mn = a
    nmn = n
print(f"O número do aluno mais alto é {nmx}, com {mx}cm;\n"
      f"O número do aluno mais baixo é {nmn}, com {mn}cm")
```
