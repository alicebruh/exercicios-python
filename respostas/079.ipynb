```python
x = int(input("Digite o ano em que estamos: "))
x = x - 1995
f = float(input("Digite seu primeiro salário: "))

for i in range(x):
  f *= (1 +(0.015*(2**i)))
print(f"Seu salário no ano atual é: R${f:.2f}")
```
