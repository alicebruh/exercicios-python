```python
n = int(input('Digite o valor de n: '))
h = 0
print(f"H = ", end="")
for i in range(n):
  print(f"1/{i+1} ", end="")
  h += 1/(i+1)
  if i < n-1:
    print("+ ", end="")
print(f"\nH = {h:.2f}")
```
