```python
idades = []
alturas = []
for i in range(5):
  idades.append(int(input("Idade da pessoa: ")))
  alturas.append(int(input("Altura da pessoa, em cm: ")))
idades.reverse()
alturas.reverse()

for i in range(5):
  print(f"A pessoa de idade {idades[i]} tem altura igual a {alturas[i]}cm.")
```
