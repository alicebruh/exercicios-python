```python
v = []
par = []
impar = []

for i in range(20):
  v.append(int(input("Digite um número inteiro: ")))
print("Elementos no vetor principal:", end="")
for i in v:
  if (i % 2 == 0):
    par.append(i)
  else:
    impar.append(i)
  print(f" {i}", end="")
print("\nElementos no vetor PAR:", end="")
for i in par:
  print(f" {i}", end="")
print("\nElementos no vetor IMPAR:", end="")
for i in impar:
  print(f" {i}", end="")
```
