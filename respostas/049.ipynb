```python
x = 0
a = 80000
b = 200000

while (a < b):
  x += 1
  a = a * ((1.03)**x)
  b = b * ((1.015)**x)

if (a == b):
  print(f"No ano {x}, as populações dos países A e B se igualaram.")
else:
  print(f"No ano {x}, a população do países A superou a população do país B.")
```
