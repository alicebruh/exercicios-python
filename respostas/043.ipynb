```python
l = float(input("Quantidade de litros: "))
c = input("Tipo de combustível: ").upper()

if (c == 'A'):
  if (l <= 20):
    p = (1.90*l)*0.97
  else:
    p = (1.90*l)*0.95
elif(c == 'G'):
  if (l <= 20):
    p = (2.50*l)*0.96
  else:
    p = (2.50*l)*0.94
print(f"O valor a ser pago é R${p:.2f}")
```
