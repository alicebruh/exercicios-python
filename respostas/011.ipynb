```python
i1 = int(input("Digite um número inteiro: "))
i2 = int(input("Digite outro número inteiro: "))
r = float(input("Digite um número real: "))

x = (i1*2)*(i2/2)
y = (3*i1)+r
z = r**3
print("O produto do dobro do primeiro com metade do segundo é {:.2f};  a soma do triplo do primeiro com o terceiro é {:.2f}; e o terceiro elevado ao cubo é {:.2f}".format(x,y,z))
```
