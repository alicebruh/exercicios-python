```python
n = int(input("Digite um número inteiro: "))
print(f"Dos números presentes no conjunto entre 1 e {n}: ")

for i in range(n-2):
  d = 0
  p = True
  for j in range(n-2):
    if (i+2 > j+1):
      d += 1
      if (((i+2)) % (j+1) == 0 and (j+1 != 1 and j+1 != (i+2))):
        p = False
    if p != True:
      break
  if p:
    print(f"O número {i+2} é primo e para encontrá-lo foi necessário fazer {d} divisão")
```
