```python
n = []
for i in range(4):
  n.append(int(input(f"Digite a nota número {i+1}: ")))
print(f"As notas digitadas foram:", end="")
for i in n:
  print(f" {i}", end="")
print(f"\nMédia das notas: {sum(n)/4}")
```
