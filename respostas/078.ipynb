```python
from math import inf
amn = pmn = inf
amx = pmx = -inf
x = y = i = 0
cpmn = camn = camx = cpmx = ""

while True:
  c = (input("Código: "))
  if (c == '0'):
    break
  i += 1
  a = float(input("Altura: "))
  x += a
  if (a > amx):
    amx = a
    camx = c
  if (a < amn):
    amn = a
    camn = c

  p = float(input("Peso: "))
  y += p
  if (p > pmx):
    pmx = p
    cpmx = c
  if (p < pmn):
    pmn = p
    cpmn = c

print(f"O cliente que tem o maior peso é o de código {cpmx}, com o peso {pmx:.2f}Kg;\n"
      f"O cliente que tem o menor peso é o de código {cpmn}, com o peso {pmn:.2f}Kg;\n"
      f"O cliente que tem a maior altura é o de código {camx}, com a altura {amx:.2f}m;\n"
      f"O cliente que tem a menor altura é o de código {camn}, com a altura {amn:.2f}m;\n"
      f"A média dos pesos é {y/i:.2f}Kg e a das alturas é {x/i:.2f}m")
```
