```python
salarios = [299,399,499,599,699,799,899,999,1000]
intervalos = [0,0,0,0,0,0,0,0,0]
while True:
  v = float(input("Escreva o valor das vendas brutas de cada vendedor, ou digite -1 para parar o programa: "))
  if (v < 0):
    break
  s = 200 + (v*0.09)
  for i in range(len(salarios)):
    if(s <= salarios[i] or (salarios[i] == 1000 and s > salarios[i])):
      intervalos[i] += 1
      break

for i in range(len(intervalos)):
  if (salarios[i] == 1000):
    print(f"A quantidade de salários no intervalo de 1000 para cima é: {intervalos[i]}")
  else:
    print(f"A quantidade de salários no intervalo entre {salarios[i]-99} e {salarios[i]}: {intervalos[i]}")
```
