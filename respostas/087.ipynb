```python
atleta = []
resultado = []

while True:
  n = input("Nome do atleta: ")
  saltos = []
  if (n == ''):
    break
  atleta.append(n)
  for i in range(5):
    s = float(input(f"Distância do salto número {i+1}: "))
    saltos.append(s)
  saltos.sort()
  m = (saltos[1]+saltos[2]+saltos[3])/3
  resultado.append(m)

  print(f"\nMelhor salto: {saltos[4]}m\n"
        f"Pior salto: {saltos[0]}m\n"
        f"Média dos demais saltos: {m:.2f}m\n")

print("Resultado Final")
for i in range(len(atleta)):
  print(f"{atleta[i]}: {resultado[i]}m")
```
