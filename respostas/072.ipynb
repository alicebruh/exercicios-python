```python
n = int(input("Quantidade de CDs: "))
t = 0
for i in range(n):
  v = float(input("Valor de cada CD: "))
  t += v
print(f"O total investido na coleção de CD é R${t:.2f}\n1"
      f"A média gasta em cada CD é R${(t/n):.2f}")
```
