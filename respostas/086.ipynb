```python
g = []
menor_acerto = 10
maior_acerto = 0
al = m = 0

for i in range(10):
  q = input(f'Digite o gabarito da questão {i+1}: ')
  g.append(q)

while True:
  ac = 0
  al += 1

  for i in range(10):
    x = input(f"Resposta questão {i+1}: ").upper()
    if (x == g[i]):
      ac += 1

  if(ac > maior_acerto):
    maior_acerto = ac
  if(ac < menor_acerto):
    menor_acerto = ac
  m += ac
  x = input("Deseja continuar usando o programa (S/N)? ")
  if (x != "S"):
    break

print(f"\nMenor número de acertos: {menor_acerto}\n"
      f"Maior número de acertos: {maior_acerto}\n"
      f"Número total de alunos: {al}\n"
      f"Média dos acertos: {m/al:.2f}")
```
