```python
n = int(input("Número de eleitores: "))
c1 = 0
c2 = 0
c3 = 0

for i in range(n):
  v = input("Selecione em qual candidato votar (A, B, ou C): ")
  if (v == 'A'):
    c1 += 1
  elif(v == 'B'):
    c2 += 1
  elif(v == 'C'):
    c3 += 1

print(f"Candidato A recebeu: {c1} votos\n"
      f"Candidato B recebeu: {c2} votos\n"
      f"Candidato C recebeu: {c3} votos")
```
