```python
# É possível inserir uma metragem com casas decimais, logo o int não atenderia a questão, pesquise por int, string, float, bool.
x = float(input("Digite um valor em metros: "))
y = x*100

# Forçe a saíde com no máximo duas casas decimais após a vírgula.
print(f"{x:.2f}m é igual a {y:.2f}cm")
```
