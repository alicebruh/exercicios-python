```python
x = float(input("Valor ganho por hora: "))
h = int(input("Número de horas trabalhadas por mês: "))
s = x*h
print("Trabalhando {} horas por mês e ganhando R${:.2f} por hora, seu salário será R${:.2f}".format(h,x,s))
```
