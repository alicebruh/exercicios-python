```python
x = float(input("Valor ganho por hora: "))

# O desafio aqui é conseguir computar as horas fracionadas, é importante utilizar
# uma medida de tempo padrão para esse caso seriam minutos. Exercitando a seguinte lógica:
# Joana trabalhou por 2 horas e 35 minutos.
# Converter as horas trabalhadas para minutos: 2 * 60 + 35,
# após isso você pode calcular o valor do minuto trabalhado e calcular o salário.
h = int(input("Número de horas trabalhadas por mês: "))
m = int(input("Minutos trabalhados há mais: "))
s = ((h*60) + m)/x
print(f"Trabalhando {((h*60) + m):.2f} minutos por mês e ganhando R${x:.2f} por hora, seu salário será R${s:.2f}")
```
