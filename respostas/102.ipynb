```python
v1 = []
v2 = []
v3 = []
v4 = []

for i in range(10):
  v1.append(input(f"Digite o elementos {i+1} do primeiro vetor: "))

for i in range(10):
  v2.append(input(f"Digite os elementos {i+1} do segundo vetor: "))

for i in range(10):
  v3.append(input(f"Digite os elementos {i+1} do terceiro vetor: "))

for i in range(10):
  v4.append(v1[i])
  v4.append(v2[i])
  v4.append(v3[i])

print(f"Quarto:", end="")
for i in v4:
  print(f" {i}", end="")
```
