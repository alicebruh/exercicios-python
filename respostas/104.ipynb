```python
temp = []
mes = ['Janeiro', 'Fevereiro', 'Março', 'Abril', 'Maio', 'Junho', 'Julho', 'Agosto', 'Setembro', 'Outubro', 'Novembro', 'Dezembro']
for i in range(12):
  temp.append(int(input(f"Temperatura média do mês número {i+1}: ")))
m = sum(temp)/12

for i in range(len(temp)):
  if temp[i] > m:
    print(f"{mes[i]}: {temp[i]} ", end="")
```
