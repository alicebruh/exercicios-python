```python
c = input("Digite o tipo da carne (F, A, P): ")
kg = float(input("Digite a quantidade de carne desjada, em Kg: "))
p = input("Forma de pagamento será cartão (C) ou dinheiro (D)? ")
d = 0

if (c == 'F'):
  t = "Filé Duplo"
  if (kg <= 5):
    x = 4.9*kg
  else:
    x = 5.8*kg
elif (c == 'A'):
  t = "Alcatra"
  if (kg <= 5):
    x = 5.9*kg
  else:
    x = 6.8*kg
elif (c == 'P'):
  t = "Picanha"
  if (kg <= 5):
    x = 6.9*kg
  else:
    x = 7.8*kg

if (p == 'C'):
  p = "Cartão"
  d = 0.05*x
else:
  p = "Dinheiro"

print(f"CUPOM FISCAL:\n"
      f"Tipo de Carne Comprada: {t}\n"
      f"Quantidade de Carne: {kg}Kg\n"
      f"Preço Total: R${x:.2f}\n"
      f"Forma de Pagamento: {p}\n"
      f"Valor do Desconto: R${d:.2f}\n"
      f"Valor a Pagar: R${(x-d):.2f}")
```
