```python
v = []
produto = 1
for i in range(5):
  v.append(int(input("Digite um número inteiro: ")))
  produto *= v[i]
print(f"A soma dos elementos desse vetor é: {sum(v)}\n"
      f"A multiplicação dos elementos desse vetor é: {produto}\n"
      "Os números dentro desse vetor são:", end="")
for i in v:
  print(f" {i}", end="")
```
